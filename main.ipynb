Building an Advanced Retrieval-Augmented Generation (RAG) Question-Answering System Using Multiple Data Sources with LangChain


```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
```

Wikipedia API Integration for Retrieval-Augmented Question Answering Using LangChain

```python
api_wrapper=WikipediaAPIWrapper(top_k_results=1,doc_content_chars_max=200)
wiki=WikipediaQueryRun(api_wrapper=api_wrapper)
```

```python
wiki.name
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
from dotenv import load_dotenv
import os
load_dotenv()
os.environ['OPENAI_API_KEY']=os.getenv('OPENAI_API_KEY')
```

```python
loader=WebBaseLoader('https://docs.smith.langchain.com/')
docs=loader.load()
documents=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200).split_documents(docs)
```

```python
## creating vector DB
vectordb=FAISS.from_documents(documents, OpenAIEmbeddings())
retriever=vectordb.as_retriever()
retriever
```

```python
from langchain.tools.retriever import create_retriever_tool
retriever_tool=create_retriever_tool(retriever,"langsmith_search","Search the connected knowledge sources for information about LangSmith. Use this tool whenever answering questions about LangSmith, and base responses on the retrieved information.")
```

```python
retriever_tool.name
```

arXiv Research Paper Search Tool

```python
from langchain_community.utilities import ArxivAPIWrapper
from langchain_community.tools import ArxivQueryRun
```

```python
arxiv_wrapper=ArxivAPIWrapper(top_k_results=1, doc_content_chars_max=200)
arxiv=ArxivQueryRun(arxiv_wrapper=arxiv_wrapper)
arxiv.name
```

Integrated Multi-Source Research Assistant Using LangChain

```python
tools=[wiki,arxiv,retriever_tool]
```

```python
tools
```

Agents in LangChain
Agents use a large language model (LLM) to decide which tools to call, what actions to take, and in what order to answer a user's query. Unlike chains, where the workflow is predefined, agents can dynamically select tools such as Wikipedia, arXiv, and FAISS retrieval based on the question.

Example: If a user asks about a recent AI research paper, the agent can choose the arXiv tool; if the question concerns information in your uploaded documents, it can use the FAISS retriever.

This enables a more flexible, intelligent, multi-source question-answering system.

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-3.5-turbo-0125", temperature=0)
```

```python
from langchain import hub
prompt = hub.pull("hwchase17/openai-functions-agent")
prompt.messages
```

```python
from langchain.agents import create_openai_tools_agent
agent=create_openai_tools_agent(llm,tools,prompt)
```

Agent Executor in LangChain

The Agent Executor manages the agent workflow by allowing the language model to choose appropriate tools execute actions process results and generate a final answer

In this project it coordinates Wikipedia arXiv and FAISS retrieval tools to answer questions using information from multiple sources

Agent decides what to do

Tools perform searches and actions

Agent Executor manages the overall process until a final answer is generated

```python
from langchain.agents import AgentExecutor
agent_executor=AgentExecutor(agent=agent,tools=tools,verbose=True)
agent_executor
```

```python
response=agent_executor.invoke({"input":"What's the paper 1605.08386 about?"})
```
